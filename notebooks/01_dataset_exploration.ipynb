```python
import re

def extract_main_category(value):
    if pd.isna(value):
        return "Unknown"

    value = str(value)

    # Remove [" and "]
    value = value.strip('["]')

    # If there is no category hierarchy, mark it unknown
    if ">>" not in value:
        return "Unknown"

    # Take the first part before >>
    return value.split(">>")[0].strip()


df["main_category"] = df["product_category_tree"].apply(extract_main_category)
```

```python
import ast

def get_first_image_url(value):
    try:
        images = ast.literal_eval(value)
        
        if isinstance(images, list) and len(images) > 0:
            return images[0]
        
        return None
    
    except:
        return None

df["image_url"] = df["image"].apply(get_first_image_url)
```

```python
target_counts = {
    "Clothing": 1000,
    "Jewellery": 700,
    "Footwear": 500,
    "Mobiles & Accessories": 450,
    "Automotive": 400,
    "Home Decor & Festive Needs": 350,
    "Beauty and Personal Care": 300,
    "Home Furnishing": 300,
    "Kitchen & Dining": 250,
    "Computers": 250,
    "Watches": 200,
    "Baby Care": 150,
    "Tools & Hardware": 150
}
```

```python
sampled_dfs = []

for category, count in target_counts.items():
    category_df = df_selected[
        df_selected["main_category"] == category
    ]
    
    sampled = category_df.sample(
        n=count,
        random_state=42
    )
    
    sampled_dfs.append(sampled)

df_final = pd.concat(sampled_dfs, ignore_index=True)

print("Final dataset size:", len(df_final))
```

```python
import requests
from pathlib import Path

test_df = df_final.head(10)

success = 0
failed = 0

for _, row in test_df.iterrows():
    pid = row["pid"]
    url = row["image_url"]

    save_path = images_dir / f"{pid}.jpg"

    try:
        response = requests.get(url, timeout=15)

        if response.status_code == 200:
            save_path.write_bytes(response.content)
            success += 1
            print(f"✅ Downloaded: {pid}")
        else:
            failed += 1
            print(f"❌ Failed: {pid} | Status: {response.status_code}")

    except Exception as e:
        failed += 1
        print(f"❌ Error: {pid} | {e}")

print("\nDownload complete")
print("Successful:", success)
print("Failed:", failed)
```

```python
import requests
from pathlib import Path
import time

# Folder for images
images_dir = Path("../data/images")
images_dir.mkdir(parents=True, exist_ok=True)

success = 0
failed = 0
skipped = 0

failed_products = []

for i, row in df_final.iterrows():

    pid = row["pid"]
    url = row["image_url"]

    save_path = images_dir / f"{pid}.jpg"

    # If already downloaded, don't download again
    if save_path.exists():
        skipped += 1
        continue

    try:
        response = requests.get(
            url,
            timeout=15,
            headers={
                "User-Agent": "Mozilla/5.0"
            }
        )

        if response.status_code == 200:
            save_path.write_bytes(response.content)
            success += 1

        else:
            failed += 1
            failed_products.append({
                "pid": pid,
                "image_url": url,
                "status": response.status_code
            })

    except Exception as e:
        failed += 1
        failed_products.append({
            "pid": pid,
            "image_url": url,
            "status": str(e)
        })

    # Progress
    if (i + 1) % 100 == 0:
        print(
            f"Processed: {i + 1}/{len(df_final)} | "
            f"Success: {success} | "
            f"Failed: {failed}"
        )

    # Small delay to avoid sending requests too quickly
    time.sleep(0.05)

print("\n========== COMPLETE ==========")
print("Successful:", success)
print("Failed:", failed)
print("Skipped:", skipped)
```

```python
ml_columns = [
    "pid",
    "product_name",
    "main_category",
    "product_category_tree",
    "brand",
    "description",
    "product_specifications",
    "retail_price",
    "discounted_price",
    "image_path"
]

df_ml = df_final[ml_columns].copy()

print("ML Dataset Shape:", df_ml.shape)
display(df_ml.head())
```

```python
print(df_ml.isnull().sum())
```

```python
# Fill missing text fields
text_columns = [
    "brand",
    "description",
    "product_specifications"
]

for col in text_columns:
    df_ml[col] = df_ml[col].fillna("")

print(df_ml.isnull().sum())
```

```python
df_ml["combined_text"] = (
    "Product: " + df_ml["product_name"] +
    " | Category: " + df_ml["main_category"] +
    " | Subcategory: " + df_ml["product_category_tree"] +
    " | Brand: " + df_ml["brand"] +
    " | Description: " + df_ml["description"] +
    " | Specifications: " + df_ml["product_specifications"]
)
```

```python
display(
    df_ml[["product_name", "combined_text"]].head(5)
)
```

```python
import re

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"\s+", " ", text)
    text = text.strip()
    return text

df_ml["combined_text_clean"] = df_ml["combined_text"].apply(clean_text)

display(
    df_ml[["product_name", "combined_text_clean"]].head(5)
)
```
